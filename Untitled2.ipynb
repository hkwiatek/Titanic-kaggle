```python
def add_features(data):
    df_out = data.copy()
    df_out['Sex'] = df_out['Sex'].map({'male':1,'female':0})
    df_out['HasCabin'] = df_out['Cabin'].notna().astype(int)
    df_out['CabinFirst'] = df_out['Cabin'].fillna('Missing').apply(lambda x: str(x)[0])

    df_out['Title'] = df_out['Name'].apply(lambda x: x.split(',')[1].split('.')[0].strip())
    df_out['Title'] = df_out['Title'].replace({'Mlle':'Miss','Ms':'Miss','Mme':'Mrs'})
    common_titles = ['Miss','Mrs','Mr','Master']
    df_out['Title'] = df_out['Title'].apply(lambda x: x if x in common_titles else 'Rare')

    df_out['FamilySize'] = df_out['SibSp'] + df_out['Parch'] + 1
    df_out['IsAlone'] = (df_out['FamilySize'] == 1).astype(int)

    df_out['FamilyCat'] = pd.cut(df_out['FamilySize'], bins = [0,1,4,20],labels=['Samotny','Mała','Duża'])
    cols_to_drop = ['PassengerId','Name','Ticket','Cabin','SibSp','Parch']
    df_out = df_out.drop(columns = [c for c in cols_to_drop if c in df_out.columns])
    return df_out
```

```python
df_train = add_features(df_train)
df_test = add_features(df_test)
#kaggle_test_processed = add_fatures(kaggle_test)
```

```python
age_medians = df_train.groupby(['Title','Pclass'])['Age'].median()
global_age = df_train['Age'].median()
freq_port = df_train['Embarked'].mode()[0]
median_fare = df_train['Fare'].median()

def clean_and_transform(df_part):
    df_clean = df_part.copy()

    def get_smart_age(row):
        if pd.isna(row['Age']):
            return age_medians.get((row['Title'],row['Pclass']),global_age)
        return row['Age']
    df_clean['Age'] = df_clean.apply(get_smart_age, axis=1)
    df_clean['IsChild'] = (df_clean['Age'] < 18).astype(int)
    df_clean['Embarked'] = df_clean['Embarked'].fillna(freq_port)
    df_clean['Fare'] = df_clean['Fare'].fillna(median_fare)
    df_clean['FarePerPerson'] = np.log1p(df_clean['Fare']/df_clean['FamilySize'])
    df_clean['Fare'] = np.log1p(df_clean['Fare'])
    return df_clean
```

```python
df_train = clean_and_transform(df_train)
df_test = clean_and_transform(df_test)
#kaggle_test_processed = clean_and_transform(kaggle_test_processed)
```

```python
X_train = df_train.drop('Survived',axis=1)
y_train = df_train['Survived']
X_test = df_test.drop('Survived',axis=1)
y_test = df_test['Survived']
```

```python
#df_train.dtypes
col_to_cod  = ['Embarked','CabinFirst','Title','FamilyCat']
X_train = pd.get_dummies(X_train,columns=col_to_cod,drop_first = True,dtype=int)
X_test = pd.get_dummies(X_test,columns=col_to_cod,drop_first = True, dtype=int)
#kaggle_test_processed = pd.get_dummies(kaggle_test_processed,columns=col_to_cod,drop_first=True,dtype=int)
X_test = X_test.reindex(columns = X_train.columns,fill_value=0)
#kaggle_test_processed = kaggle_test_processed.reindex(columns=X_train.columns,fill_value=0)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
#kaggle_test_scaled = scaler.transform(kaggle_test_processed)
```

```python
param_grids = {
    'Logistic Regression':(LogisticRegression(random_state = 42,max_iter=1000),{'C':[0.05,0.1,0.5,1.0,5.0]}),
    'Random Forest': (RandomForestClassifier(random_state=42),{'n_estimators':[100,200,400],'max_depth':[4,6,8,None],
                                                               'min_samples_split':[2,5,10],'min_samples_leaf':[1,2,4]}),
    'XGBoost':(XGBClassifier(random_state=42,eval_metric='logloss'),{'n_estimators':[100,200,500],'max_depth':[3,4,5],'learning_rate':[0.01,0.05,0.1],
                                                                 'subsample':[0.8,1.0],'colsample_bytree':[0.8,1.0]})}
best_models={}
cv_scores = {}
for name, (model,grid) in param_grids.items():
    search = GridSearchCV(estimator = model, param_grid = grid, cv=8, scoring='accuracy',n_jobs=-1)
    search.fit(X_train_scaled,y_train)
    best_models[name] = search.best_estimator_
    cv_scores[name] = search.best_score_
    print(f"Model {name} accuracy na df_train {search.best_score_:.4f} | Parametry: {search.best_params_}")
```

```python
voting_clf = VotingClassifier(estimators = [
        ('lr', best_models['Logistic Regression']),
        ('rf',best_models['Random Forest']),
        ('xgb',best_models['XGBoost'])],voting = 'soft')
voting_clf.fit(X_train_scaled,y_train)
best_models['Voting Ensemble'] = voting_clf
cv_scores['Voting Ensemble'] = cross_val_score(
    voting_clf, X_train_scaled, y_train, cv=8, scoring='accuracy', n_jobs=-1
).mean()
```

```python
wyniki_tabela = []

for name, model in best_models.items():
    test_preds = model.predict(X_test_scaled)
    test_acc = accuracy_score(y_test, test_preds)
    
    wyniki_tabela.append({
        'Model': name,
        'CV Accuracy (Train)': round(cv_scores.get(name, np.nan), 4),
        'Test Accuracy (df_test)': round(test_acc, 4)
    })

display(pd.DataFrame(wyniki_tabela))
```
